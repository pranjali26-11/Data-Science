difference between list and numpy array:

list is built in | numpy is external lib

list dosent support mathematical operation| support mathematical operation

list contain values with diff dtype| numpy array contain same dtype element

dont support multidimensionality or it is messy| support 2D 3D multidimensional array

less powerful | more powerful








**data type hierarchy**:

precedence to data type:

string(object)>complex>float>int

if we have change any value of int array to float then all array values will convert to float


```python
#value converted to float . float>int
np.array([10,20,30,40.])
```

**bold text**

```python
#all float values converted to complex values
np.array([10,20,(30+4j),40.])
```

```python
#all values are converted to strings
c = np.array([10,20,(30+4j),40.,'A'])
```
