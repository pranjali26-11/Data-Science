

>**1.['estimated_holiday_spending'] (Bracket notation):**
Accesses a column/key using its exact name as a string
Works always, even if:
1.Column name has spaces ("holiday spending")
2.Column name has special characters ($, -, etc.)
3.Column name conflicts with existing methods


**2. s.estimated_holiday_spending (Dot notation)**
Accesses column like an attribute
Works only if:
Name has no spaces or special characters
Name does not conflict with built-in attributes/methods
